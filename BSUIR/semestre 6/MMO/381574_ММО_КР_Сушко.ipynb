```python
from sklearn.datasets import load_breast_cancer
import pandas as pd

# Загрузка датасета
bc_data = load_breast_cancer(as_frame=True)

# Создание DataFrame для признаков и целевой переменной
df = bc_data.frame

# Вывод первых 5 строк датасета
display(df.head())
```

```python
# Вывод информации о датасете
display(df.info())
```

В этом наборе данных `target` - это целевая переменная, где `0` соответствует злокачественной опухоли (malignant), а `1` - доброкачественной (benign). Признаки описывают характеристики клеточных ядер.

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Выделение признаков (X) и целевой переменной (y)
X = df.drop('target', axis=1)
y = df['target']

# Разделение данных на обучающую и тестовую выборки
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# Стандартизация признаков
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Преобразование стандартизированных данных обратно в DataFrame для удобства анализа (опционально)
X_train_scaled = pd.DataFrame(X_train_scaled, columns=X.columns)
X_test_scaled = pd.DataFrame(X_test_scaled, columns=X.columns)

display(X_train_scaled.head())
```

Мы успешно стандартизировали признаки и разделили данные. Теперь можно приступить к созданию и обучению моделей классификации.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

# Инициализация моделей
models = {
    'Logistic Regression': LogisticRegression(random_state=42, solver='liblinear'),
    'SVM': SVC(random_state=42, probability=True), # probability=True для roc_auc_score
    'CART': DecisionTreeClassifier(random_state=42)
}

# Обучение и оценка моделей
results = {}

for name, model in models.items():
    model.fit(X_train_scaled, y_train)
    y_pred = model.predict(X_test_scaled)
    y_proba = model.predict_proba(X_test_scaled)[:, 1] # Вероятности для ROC AUC

    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    roc_auc = roc_auc_score(y_test, y_proba)

    results[name] = {
        'Accuracy': accuracy,
        'Precision': precision,
        'Recall': recall,
        'F1-score': f1,
        'ROC AUC': roc_auc
    }

# Вывод результатов
results_df = pd.DataFrame(results).T
display(results_df)
```

# Оценки качества
- Logistic Regression показала наилучшие результаты по всем метрикам: Accuracy, Precision, Recall, F1-score и ROC AUC близки к 0.99.
- SVM также продемонстрировала высокую производительность, немного уступая логистической регрессии.
- CART (Decision Tree) показала заметно более низкие результаты по сравнению с другими моделями, особенно по ROC AUC, что может указывать на переобучение или меньшую обобщающую способность на этом датасете без дополнительной настройки.

# Выводы
### Логистическая регрессия (Logistic Regression)
Моделирует вероятность того, что объект принадлежит к определенному классу, используя сигмоидную функцию.

Плюсы:
- Высокая интерпретируемость (веса показывают важность и направление влияния признака).
- Выдает калиброванные вероятности.
- Работает быстро, легко обновляется онлайн.

Минусы:
- Не может выучить нелинейные зависимости без ручного создания новых признаков (полиномиальных).
- Чувствительна к выбросам (выбросы могут сильно «наклонить» границу).
- Требует масштабирования признаков (feature scaling).

### Метод опорных векторов (SVM)
Находит гиперплоскость, которая не просто разделяет классы, а максимизирует зазор (margin) между ближайшими точками разных классов (опорными векторами).

Плюсы:
- Очень эффективен в пространствах высокой размерности (когда признаков больше, чем объектов).
- Устойчив к переобучению в высоких размерностях благодаря максимизации зазора.
Менее чувствителен к выбросам, чем LR (важны только опорные векторы на границе).

Минусы:
- Плохо работает на больших датасетах (медленное обучение).
- Требует тщательного подбора ядра и параметров.
- Требует масштабирования признаков.
- «Черный ящик» при использовании нелинейных ядер (трудно интерпретировать).

### CART (Classification and Regression Trees)
Это «классическое» дерево решений (алгоритм, лежащий в основе Random Forest и Gradient Boosting). Рекурсивно разбивает пространство признаков на прямоугольные области, задавая пороги.

Плюсы:
- Не требует масштабирования данных (работает с разными масштабами).
- Легко интерпретировать (можно нарисовать дерево и понять логику).
- Автоматически улавливает нелинейности и взаимодействия признаков.
- Работает с категориальными и численными данными.

Минусы:
- Склонен к сильному переобучению (нужна обрезка/pruning).
- Нестабилен: небольшое изменение данных может полностью изменить структуру дерева (высокая дисперсия).
- Не может экстраполировать (предсказывать значения за пределами обучающей выборки в регрессии).
