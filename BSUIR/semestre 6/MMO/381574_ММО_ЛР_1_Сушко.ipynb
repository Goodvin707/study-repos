# Задание 1. Загрузка и первичный анализ данных
Загрузить датасет ирисов Фишера из библиотеки scikit-learn, преобразовать его в pandas DataFrame, вывести первые строки. Выполнить первичный анализ данных: получить общую информацию о датасете (типы данных, пропущенные значения), рассчитать базовую статистику для числовых признаков (среднее, стандартное отклонение, мин/макс). Проанализировать распределение классов для проверки сбалансированности и описать физический смысл каждого признака. Подвести итоги первичного анализа данных.

````python
import pandas as pd
from sklearn.datasets import load_iris

# Загрузка датасета ирисов
iris_data = load_iris()

# Создание pandas DataFrame
df = pd.DataFrame(data=iris_data.data, columns=iris_data.feature_names)

# Добавление целевой переменной (вида ириса) в DataFrame
df['target'] = iris_data.target

# Вывод первых нескольких строк данных для понимания их структуры
print("Первые 5 строк датасета ирисов:")
display(df.head())

# Получение общей информации о датасете
print("\nОбщая информация о датасете:")
df.info()

# Проверка на наличие пропущенных значений
print("\nКоличество пропущенных значений в каждом столбце:")
print(df.isnull().sum())

# Расчет базовой статистики по всем числовым признакам
print("\nБазовая статистика по числовым признакам:")
display(df.describe())
````

````python
print("Среднее значение по числовым признакам:\n", df.describe().loc['mean'])
print("\nСтандартное отклонение по числовым признакам:\n", df.describe().loc['std'])
print("\nМинимальные значения по числовым признакам:\n", df.describe().loc['min'])
print("\nМаксимальные значения по числовым признакам:\n", df.describe().loc['max'])
````

````python
import pandas as pd
from sklearn.datasets import load_iris

# Загрузка целевой переменной для получения названий классов
iris_data = load_iris()
target_names = iris_data.target_names

# Анализ распределения классов
print("\nРаспределение классов (целевая переменная):")
class_distribution = df['target'].value_counts().sort_index()
print(class_distribution)

print("\nРаспределение классов по видам ирисов:")
for i, count in enumerate(class_distribution):
    print(f"Класс {i} ({target_names[i]}): {count} образцов")

# Проверка на сбалансированность
if len(class_distribution.unique()) == 1:
    print("\nДанные сбалансированы, каждый класс содержит одинаковое количество образцов.")
else:
    print("\nДанные несбалансированы, количество образцов в классах отличается.")
````

### Описание признаков и их физический смысл

Датасет ирисов Фишера содержит четыре числовых признака, которые описывают морфологические характеристики цветков ириса, и одну целевую переменную, представляющую вид ириса.

*   **sepal length (cm)**: Длина чашелистика в сантиметрах. Чашелистики — это внешние лепестковидные структуры, которые окружают и защищают бутон цветка до его распускания. Этот признак помогает различать виды ирисов, так как их чашелистики могут значительно отличаться по размеру.
*   **sepal width (cm)**: Ширина чашелистика в сантиметрах. Наряду с длиной чашелистика, ширина также является важной морфологической характеристикой, которая может варьироваться между видами и использоваться для их идентификации.
*   **petal length (cm)**: Длина лепестка в сантиметрах. Лепестки — это внутренние, часто ярко окрашенные структуры цветка, привлекающие опылителей. Длина лепестка обычно является одним из наиболее выраженных признаков для разделения видов ирисов, особенно между _Iris setosa_ и остальными.
*   **petal width (cm)**: Ширина лепестка в сантиметрах. Ширина лепестка также является критически важным признаком для классификации, часто демонстрируя четкие различия между видами, например, у _Iris virginica_ обычно самые широкие лепестки.
*   **target**: Целевая переменная, представляющая вид ириса. Она закодирована числами: 0 для _Iris setosa_, 1 для _Iris versicolor_ и 2 для _Iris virginica_. Это категориальный признак, который мы хотим предсказать, используя остальные четыре числовых признака.

# Задание 3. Разделение данных и нормализация

Подготовьте данные для обучения моделей машинного обучения.
Разделите исходный датасет на обучающую и тестовую выборки в соотношении восемьдесят к двадцати процентам. Используйте стратифицированное разделение, чтобы сохранить пропорции классов в обеих выборках. Зафиксируйте случайное состояние для воспроизводимости результатов. Выполните стандартизацию признаков, применив стандартный скейлер. Важно обучить скейлер только на обучающей выборке, а затем применить его как к обучающей, так и к тестовой выборке. Объясните, почему нормализация важна для некоторых алгоритмов машинного обучения и почему скейлер должен обучаться только на обучающих данных.

````python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Разделение признаков (X) и целевой переменной (y)
X = df.drop('target', axis=1)
y = df['target']

# Разделение датасета на обучающую и тестовую выборки
# test_size=0.2 означает 20% данных для теста, 80% для обучения
# stratify=y обеспечивает стратифицированное разбиение, сохраняя пропорции классов
# random_state для воспроизводимости результатов
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

print("Размер обучающей выборки (X_train):", X_train.shape)
print("Размер тестовой выборки (X_test):", X_test.shape)
print("Распределение классов в обучающей выборке:\n", y_train.value_counts(normalize=True))
print("Распределение классов в тестовой выборке:\n", y_test.value_counts(normalize=True))

# Инициализация StandardScaler
scaler = StandardScaler()

# Обучение скейлера только на обучающей выборке и преобразование обучающей выборки
X_train_scaled = scaler.fit_transform(X_train)

# Применение (трансформация) обученного скейлера к тестовой выборке
X_test_scaled = scaler.transform(X_test)

print("\nПервые 5 строк стандартизированной обучающей выборки (X_train_scaled):")
print(X_train_scaled[:5])

print("\nПервые 5 строк стандартизированной тестовой выборки (X_test_scaled):")
print(X_test_scaled[:5])
````

## Важность нормализации (стандартизации) для некоторых алгоритмов машинного обучения:

- Алгоритмы, основанные на расстоянии: Многие алгоритмы, такие как K-ближайшие соседи (KNN), метод опорных векторов (SVM) и кластеризация K-Means, измеряют расстояние между точками данных. Если признаки имеют разные масштабы, признаки с большими значениями будут доминировать при расчете расстояния. Нормализация гарантирует, что все признаки вносят пропорциональный вклад.
- Градиентные алгоритмы: Алгоритмы, использующие градиентный спуск (например, логистическая регрессия, нейронные сети), сходятся быстрее, когда признаки находятся в одном масштабе. Разные масштабы могут привести к тому, что функция потерь будет иметь очень вытянутую форму, затрудняя нахождение глобального минимума.
- Регуляризация: Методы регуляризации, такие как L1 (Lasso) и L2 (Ridge), штрафуют большие веса. Если признаки не нормализованы, алгоритм может присвоить большие веса признакам с большим масштабом, несправедливо наказывая их или наоборот, игнорируя менее масштабные, но важные признаки. Нормализация обеспечивает справедливое применение регуляризации.

## Почему скейлер должен обучаться только на обучающих данных:

Обучение скейлера (например, StandardScaler) на всем датасете (обучающей и тестовой выборках вместе) приводит к утечке данных (data leakage). Это означает, что информация из тестовой выборки (например, среднее значение или стандартное отклонение) косвенно попадает в обучающий процесс. В результате модель будет иметь более оптимистичную оценку производительности на тестовой выборке, чем она была бы в реальных условиях, где модель сталкивается с совершенно новыми, невиданными ранее данными.

Правильный подход заключается в следующем:

Обучить скейлер на X_train: Рассчитать параметры преобразования (среднее, стандартное отклонение и т.д.) только на обучающих данных (X_train).
Применить преобразование к X_train и X_test: Использовать те же параметры преобразования, полученные на X_train, для стандартизации как обучающей, так и тестовой выборок. Это имитирует реальный сценарий, когда модель, обученная на исторических данных, применяется к новым, будущим данным.

Точность (Accuracy) — это доля правильных прогнозов модели среди всех сделанных прогнозов.

```
Accuracy = (TP + TN) / (TP + TN + FP + FN)
```

Precision (Точность/Прецизионность) — качество положительных прогнозов. Какая доля объектов, которые модель назвала положительными (1), действительно является положительными?

`Precision = TP / (TP + FP)`

Recall (Полнота) — полнота охвата положительного класса. Какая доля реально положительных объектов была успешно обнаружена моделью?

`Recall = TP / (TP + FN)`

F1-мера (F1-score) — гармоническое среднее между Precision и Recall. Это единая метрика, которая балансирует оба аспекта. Она высока только когда высоки и Precision, и Recall.

`F1 = 2 * (Precision * Recall) / (Precision + Recall)`

2. Логистическая регрессия: принцип работы и парадокс названия
Почему это "регрессия", хотя используется для классификации?
Исторически и математически логистическая регрессия — это разновидность регрессионного анализа. Она моделирует не саму категорию, а вероятность принадлежности к классу (зависимую непрерывную переменную в диапазоне [0, 1]). То есть она регрессирует вероятность. Название отражает метод, а не конечную задачу.

Принцип работы:

Линейная комбинация: Как и в линейной регрессии, модель вычисляет взвешенную сумму признаков объекта: z = w0 + w1*x1 + w2*x2 + ... + wn*xn, где w — веса (коэффициенты), x — признаки.

Сигмоидная функция (логистическая функция): Этот линейный результат z пропускается через сигмоиду (σ(z) = 1 / (1 + e^{-z})). Эта функция "сжимает" любое число z в интервал между 0 и 1, что можно интерпретировать как вероятность P(y=1 | x).

Полученная вероятность сравнивается с порогом (по умолчанию 0.5). Если P > 0.5, объект относится к классу 1, иначе — к классу 0.

Цель обучения — найти такие веса w, чтобы максимизировать функцию правдоподобия (или минимизировать логарифмическую функцию потерь), т.е. сделать прогнозируемые вероятности как можно ближе к реальным меткам.

# Задание 5
Примените алгоритм k-ближайших соседей для классификации. Подберите оптимальное значение гиперпараметра k, количества ближайших соседей, которое следует учитывать при классификации. Протестируйте значения k от одного до двадцати. Для каждого значения k выполните кроссвалидацию на обучающей выборке с пятью фолдами и рассчитайте среднюю точность. Постройте график зависимости точности от значения k. Определите оптимальное k, при котором достигается максимальная точность на кроссвалидации. Обучите финальную модель с лучшим значением k и оцените ее на тестовой выборке. Объясните принцип работы алгоритма k-ближайших соседей, его преимущества и недостатки. Обсудите, почему для этого алгоритма критически важна нормализация признаков.

````python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score
from sklearn.metrics import classification_report
import numpy as np
import matplotlib.pyplot as plt

# Список для хранения средних точностей для разных k
k_values = list(range(1, 21))
mean_accuracies = []

# Перебор значений k от 1 до 20
for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    # Выполнение 5-фолдовой кросс-валидации на обучающей выборке
    scores = cross_val_score(knn, X_train_scaled, y_train, cv=5, scoring='accuracy')
    # Расчет средней точности и добавление ее в список
    mean_accuracies.append(scores.mean())

# Вывод средних точностей для каждого k
print("Средние точности для каждого значения k:")
for k, acc in zip(k_values, mean_accuracies):
    print(f"k = {k}: {acc:.4f}")

# Построение графика зависимости точности от значения k
plt.figure(figsize=(10, 6))
plt.plot(k_values, mean_accuracies, marker='o', linestyle='-')
plt.title('Зависимость средней точности от значения k для KNN (кросс-валидация)')
plt.xlabel('Количество соседей (k)')
plt.ylabel('Средняя точность кросс-валидации')
plt.xticks(k_values)
plt.setp(plt.scatter(5, 0.9667), color='r', linewidth=5.0)
plt.grid(True)
plt.show()

# Определение оптимального k
optimal_k_index = np.argmax(mean_accuracies)
optimal_k = k_values[optimal_k_index]
max_accuracy = mean_accuracies[optimal_k_index]

print(f"\nОптимальное значение k: {optimal_k} (с точностью: {max_accuracy:.4f})")

# Обучение финальной модели с лучшим значением k
final_knn_model = KNeighborsClassifier(n_neighbors=optimal_k)
final_knn_model.fit(X_train_scaled, y_train)

# Оценка модели на тестовой выборке
y_pred_knn = final_knn_model.predict(X_test_scaled)

print("\nОтчет по классификации для KNN с оптимальным k:")
print(classification_report(y_test, y_pred_knn, target_names=iris_data.target_names, zero_division=0))
````

### Принцип работы алгоритма k-ближайших соседей
k-NN — один из самых простых и интуитивно понятных алгоритмов машинного обучения, основанный на принципе подобия: "скажи мне, кто твои соседи, и я скажу, кто ты".

Как работает (в фазе предсказания):

1. Запоминание: На этапе обучения алгоритм просто запоминает все обучающие данные (нормально, без построения явной модели). Это делает его "ленивым" алгоритмом (lazy learner).
2. Предсказание для нового объекта:
- Вычисляются расстояния от нового объекта до всех объектов в обучающей выборке.
- Выбираются k ближайших объектов (с наименьшими расстояниями).
- Для классификации: определяется класс по мажоритарному голосованию — какой класс чаще встречается среди k соседей.
- Для регрессии: вычисляется среднее значение целевой переменной среди k соседей.

### Преимущества:
- Простота реализации и понимания — интуитивно ясная логика.
- Не делает предположений о распределении данных — непараметрический метод.
- Универсальность — работает как для классификации, так и для регрессии.
- Хорошо работает на небольших наборах данных с четкой кластерной структурой.

- Интерпретируемость — можно показать, на основе каких примеров было принято решение.

### Недостатки:
- Вычислительная сложность — на предсказание нужно вычислять расстояния до всех объектов обучающей выборки. Сложность O(n) для каждого предсказания.
- Чувствительность к размерности данных ("проклятие размерности") — эффективность резко падает при большом количестве признаков.
- Требует хранения всей обучающей выборки — может занимать много памяти.
- Чувствительность к нерелевантным признакам и шуму в данных.

````python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# Инициализация и обучение SVM с линейным ядром
svm_linear = SVC(kernel='linear', random_state=42)
svm_linear.fit(X_train_scaled, y_train)
y_pred_linear = svm_linear.predict(X_test_scaled)
accuracy_linear = accuracy_score(y_test, y_pred_linear)
print(f"Точность SVM с линейным ядром: {accuracy_linear:.4f}")

# Инициализация и обучение SVM с радиальным базисным ядром (RBF)
svm_rbf = SVC(kernel='rbf', random_state=42)
svm_rbf.fit(X_train_scaled, y_train)
y_pred_rbf = svm_rbf.predict(X_test_scaled)
accuracy_rbf = accuracy_score(y_test, y_pred_rbf)
print(f"Точность SVM с RBF ядром: {accuracy_rbf:.4f}")

# Инициализация и обучение SVM с полиномиальным ядром
# Можно настроить степень полинома (degree) и коэффициент (coef0), но начнем с дефолтных
svm_poly = SVC(kernel='poly', random_state=42)
svm_poly.fit(X_train_scaled, y_train)
y_pred_poly = svm_poly.predict(X_test_scaled)
accuracy_poly = accuracy_score(y_test, y_pred_poly)
print(f"Точность SVM с полиномиальным ядром: {accuracy_poly:.4f}")

# Сравнение результатов
print("\nСравнение точности моделей SVM с разными ядрами:")
print(f"  Линейное ядро: {accuracy_linear:.4f}")
print(f"  RBF ядро:      {accuracy_rbf:.4f}")
print(f"  Полиномиальное ядро: {accuracy_poly:.4f}")

# Определение лучшего ядра
best_kernel = ''
max_accuracy = 0

if accuracy_linear > max_accuracy:
    max_accuracy = accuracy_linear
    best_kernel = 'Линейное'
if accuracy_rbf > max_accuracy:
    max_accuracy = accuracy_rbf
    best_kernel = 'RBF'
if accuracy_poly > max_accuracy:
    max_accuracy = accuracy_poly
    best_kernel = 'Полиномиальное'

print(f"\nЛучшее ядро для данной задачи: {best_kernel} с точностью {max_accuracy:.4f}")
````

Ядро в SVM — это трюк для неявного преобразования данных в пространство большей размерности, где они становятся линейно разделимыми, без реального вычисления этого преобразования.
Как работает:

Проблема: Данные не разделимы прямой в исходном пространстве (например, концентрические круги)

Решение: Мысленно "поднимаем" данные в 3D (добавляем новый признак z = x² + y²)

В 3D данные становятся разделимы плоскостью

Ядро вычисляет скалярные произведения в этом новом пространстве, не строя его явно

Математически: SVM нужно только K(x₁, x₂) = ⟨φ(x₁), φ(x₂)⟩, а не сами φ(x)

Типы ядер и их применение:
1. Линейное — K(x₁, x₂) = x₁·x₂
Когда: Много признаков (>1000), данные почти линейно разделимы

Плюс: Быстро, нет лишних параметров

Пример: Текстовые данные (Bag of Words)

2. Полиномиальное — K(x₁, x₂) = (γ·x₁·x₂ + r)ᵈ
Когда: Известно, что зависимость полиномиальная

Контроль: Степенью d управляем сложностью

Минус: Склонен к переобучению при больших d

3. RBF (Гауссово) — K(x₁, x₂) = exp(-γ·||x₁ - x₂||²) ★ САМОЕ ПОПУЛЯРНОЕ
Когда: Нет априорных знаний о данных

Гибкость: Может аппроксимировать любую границу

Параметр: γ — чем больше, тем сложнее граница

Де-факто выбор по умолчанию

4. Сигмоидное — K(x₁, x₂) = tanh(γ·x₁·x₂ + r)
Когда: Данные напоминают работу нейронной сети

Редко используется на практике

Почему SVM эффективен при малом числе признаков:
1. Работает с границей, а не объемом
k-NN страдает в многомерности ("все точки далеки друг от друга")

SVM ищет только разделяющую гиперплоскость

Использует опорные векторы (обычно малая часть данных)

2. Maximum Margin = встроенная регуляризация
Широкий зазор между классами предотвращает переобучение

Особенно важно при малом количестве данных

3. Ядра создают "виртуальные" признаки
RBF-ядро неявно работает в бесконечномерном пространстве

Но делает это эффективно через ядро

4. Проблемы SVM при многих признаках:
Линейное ядро: O(n_features) — нормально

Неlinear ядра: O(n_samples² × n_features) — медленно

Память для матрицы ядра: O(n_samples²) — много
